## 2. 거리 기반 분류

두 개의 특성(Feature)으로 표현된 학습 샘플 네 개와 새 입력의 거리를 계산합니다. 가장 가까운 샘플의 레이블(Label)을 예측 결과로 사용합니다.


```python
import numpy as np

samples = np.array([[2, 1], [3, 2], [7, 7], [8, 6]], dtype=float)
labels = np.array(["A", "A", "B", "B"])
query = np.array([4, 3], dtype=float)

distances = np.linalg.norm(samples - query, axis=1)
nearest_index = int(np.argmin(distances))
print("거리:", np.round(distances, 3))
print("가장 가까운 사례:", nearest_index + 1)
print("예측:", labels[nearest_index])
```

## 3. Iris 분류

붓꽃의 측정값으로 품종을 분류하는 최근접 이웃(KNN) 모델을 만듭니다. `fit()`으로 모델을 학습하고 `predict()`로 평가 데이터의 레이블을 예측합니다. 표준화에 필요한 값은 학습 데이터에서 계산합니다.


```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.2, random_state=42, stratify=iris.target
)
model = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=5))
model.fit(X_train, y_train)
predicted = model.predict(X_test)
print(f"평가: {len(y_test)}건 / 정답: {(predicted == y_test).sum()}건")
print(f"정확도: {(predicted == y_test).mean():.1%}")
```

## 4. 예측 결과 시각화

그래프에는 꽃잎 길이와 너비만 표시하지만 모델은 네 가지 Feature를 모두 사용합니다. 검은색 ×는 오분류된 샘플입니다.


```python
import matplotlib.pyplot as plt

wrong = predicted != y_test
fig, ax = plt.subplots(figsize=(7, 4.5))
scatter = ax.scatter(X_test[:, 2], X_test[:, 3], c=predicted, cmap="viridis")
if wrong.any():
    ax.scatter(X_test[wrong, 2], X_test[wrong, 3], c="black", marker="x", s=110, label="Incorrect")
    ax.legend()
ax.set(xlabel="Petal length (cm)", ylabel="Petal width (cm)", title="Iris test predictions")
fig.colorbar(scatter, ax=ax, ticks=[0, 1, 2], label="Predicted class")
fig.tight_layout()
plt.show()
```

## 5. Accuracy의 한계

전체 100건 중 정상 90건, 드문 사건 10건인 데이터를 가정합니다. 모두 정상으로 분류하면 정확도(Accuracy)는 90%이지만 드문 사건의 재현율(Recall)은 0%입니다.


```python
y_true = np.array([0] * 90 + [1] * 10)
y_pred = np.zeros(100, dtype=int)
accuracy = (y_true == y_pred).mean()
rare_mask = y_true == 1
rare_recall = (y_pred[rare_mask] == 1).mean()
print(f"정확도: {accuracy:.1%}")
print(f"드문 사건 재현율: {rare_recall:.1%}")
```
